```python
print(df)
```

Q1. How many employees are there in each department?

```python
cnt=df['Department'].value_counts()
print(cnt)
```

2. What is the average salary of all employees?

```python
avgsal=df['Salary'].sum()/len(df)
print(f"Average salary of Employee : {avgsal:.2f} ")
```

3. What is the minimum and maximum salary in the dataset?

```python
minn=df['Salary'].min()
maxx=df['Salary'].max()
print(f'Minimum salary is : {minn:.2f}')
print(f'Maximum salary is : {maxx:.2f}')
```

4. How many employees hold each designation (Junior, Senior, Manager)?

```python
cnntt=df['Designation'].value_counts()
print(cnntt)
```

5. Count the number of employees by gender.

```python
gcnt=df['Gender'].value_counts()
print(gcnt)
```

6. How many employees joined after the year 2020?

```python
df['Date of Joining']=pd.to_datetime(df['Date of Joining'])
after_2020=df[df['Date of Joining'].dt.year>2020]
countt=after_2020.shape[0]

print(countt)
```

7. What is the average age of employees?

```python
df['Date of Birth']=pd.to_datetime(df['Date of Birth'])
today=pd.to_datetime("today")
df['age']=(today-df['Date of Birth']).dt.days//365
average_age=df['age'].mean()

print(f"Average age of employees {average_age:.2f} years")
```

8. Which year had the most employee joinings?

```python
df['Date of Joining']=pd.to_datetime(df['Date of Joining'])
df['joining year']=df['Date of Joining'].dt.year

yearly_counts=df['Date of Joining'].value_counts().sort_index(ascending=True)

most_joining_year=yearly_counts.idxmax()
most_num_employee=yearly_counts.max()

print(f"The most people join in {most_joining_year} with {most_num_employee} people")
```

9. Which department has the highest average salary?

```python
dept_avg_salary=df.groupby('Department')['Salary'].mean()

top_dept=dept_avg_salary.idxmax()
top_sal=dept_avg_salary.max()

print(f"Top department with average salary is {top_dept} with {top_sal:.2f} per employee")
```
